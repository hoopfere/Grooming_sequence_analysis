```python
import mat73
import pandas as pd
import seaborn as sns
```

```python
data_dict = mat73.loadmat("trx.mat")
```

```python
data_dict['timestamps'].keys()
```

```python
coords = pd.DataFrame.from_dict(data_dict['trx'])[["x", "y"]].explode(["x", "y"]).reset_index(drop=True)
```

```python
coords
```

```python
sns.lineplot(coords, x="x", y="y", linewidth=0.5, sort=False)
```

```python
sns.kdeplot(coords, x="x", y="y", fill=True, thresh=0, levels=100)
```
