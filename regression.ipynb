```python
# imports

from MarkovChain import MarkovChain
import pandas as pd
import statsmodels.api as sm
```

```python
gh_markov = MarkovChain("final_all_gh.csv")
sh_markov = MarkovChain("final_all_sh.csv")

genotype_list = ["group" for _ in gh_markov.frequency_list] + ["solo" for _ in sh_markov.frequency_list]
```

```python
X = sm.add_constant(pd.DataFrame([i.stack().to_frame()[0] for i in gh_markov.frequency_list] + [i.stack().to_frame()[0] for i in sh_markov.frequency_list]))
Y = [int(i == "solo") for i in ["group" for _ in gh_markov.frequency_list] + ["solo" for _ in sh_markov.frequency_list]]
```

```python
logit_mod = sm.Logit(Y, X)
logit_res = logit_mod.fit()
logit_res.summary()
```
